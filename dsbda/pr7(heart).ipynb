```python
# Import pandas
import pandas as pd

# Import matplotlib
import matplotlib.pyplot as plt

# Import seaborn
import seaborn as sns
```

```python
# Load heart dataset
heart = pd.read_csv("heart.csv")
```

```python
# Display first 5 rows
print(heart.head())
```

```python
# Histogram

plt.hist(heart["Age"])

plt.title("Age Histogram")

plt.xlabel("Age")

plt.ylabel("Frequency")

plt.show()
```

```python
# Scatter plot

plt.scatter(heart["Age"], heart["Chol"])

plt.title("Age vs Cholesterol")

plt.xlabel("Age")

plt.ylabel("Cholesterol")

plt.show()
```

```python
# Bar plot

sns.barplot(x=heart["Sex"], y=heart["Chol"])

plt.title("Sex vs Cholesterol")

plt.show()
```

```python
# Heatmap

sns.heatmap(heart.corr(numeric_only=True), annot=True)

plt.title("Heart Disease Heatmap")

plt.show()
```
