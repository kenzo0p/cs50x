```python
# Import pandas library
import pandas as pd

# Import numpy library
import numpy as np
```

```python
# Load forest fire dataset
forest = pd.read_csv("forestfires.csv")
```

```python
# Display first 5 rows
print(forest.head())
```

```python
# Display dataset information
print(forest.info())
```

```python
# Display column names
print(forest.columns)
```

```python
# Check missing values
print(forest.isnull().sum())
```

```python
# Remove missing values
forest = forest.dropna()

# Display dataset
print(forest.head())
```

```python
# Remove duplicate rows
forest = forest.drop_duplicates()

# Display dataset
print(forest.head())
```

```python
# Mapper Step

temp_list = []

for temp in forest['temp']:
    temp_list.append(temp)

print(temp_list[:10])
```

```python
# Reducer Step

total = 0
count = 0

for value in temp_list:
    total = total + value
    count = count + 1

average = total / count

print("Average Temperature:", average)
```

```python
# Find maximum wind speed
print("Maximum Wind Speed:", forest['wind'].max())
```

```python
# Find total burned area
print("Total Burned Area:", forest['area'].sum())
```

```python
# Average temperature month wise
print(forest.groupby('month')['temp'].mean())
```

```python
# Sort temperature values
print(forest.sort_values(by='temp').head())
```

```python
forest.to_csv("forestfire_output.csv", index=False)

print("Output file generated successfully")
```
