```python
# Import pandas
import pandas as pd

# Import matplotlib
import matplotlib.pyplot as plt

# Import seaborn
import seaborn as sns
```

```python
# Load forest fire dataset
forest = pd.read_csv("forestfires.csv")
```

```python
# Display first 5 rows
print(forest.head())
```

```python
# Line plot

plt.plot(forest["temp"])

plt.title("Temperature Line Plot")

plt.xlabel("Index")

plt.ylabel("Temperature")

plt.show()
```

```python
# Bar plot

sns.barplot(x=forest["month"], y=forest["temp"])

plt.title("Month vs Temperature")

plt.show()
```

```python
# Histogram

plt.hist(forest["wind"])

plt.title("Wind Histogram")

plt.xlabel("Wind")

plt.ylabel("Frequency")

plt.show()
```

```python
# Scatter plot

plt.scatter(forest["temp"], forest["area"])

plt.title("Temperature vs Area")

plt.xlabel("Temperature")

plt.ylabel("Area")

plt.show()
```

```python
# Heatmap

sns.heatmap(forest.corr(numeric_only=True), annot=True)

plt.title("Forest Fire Heatmap")

plt.show()
```
