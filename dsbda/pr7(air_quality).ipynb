```python
# Import pandas
import pandas as pd

# Import matplotlib
import matplotlib.pyplot as plt

# Import seaborn
import seaborn as sns
```

```python
# Load air quality dataset
air = pd.read_csv("airquality.csv")
```

```python
# Display first 5 rows
print(air.head())
```

```python
# Line plot

plt.plot(air["Ozone"])

plt.title("Ozone Line Plot")

plt.xlabel("Index")

plt.ylabel("Ozone")

plt.show()
```

```python
# Histogram

plt.hist(air["Temp"])

plt.title("Temperature Histogram")

plt.xlabel("Temperature")

plt.ylabel("Frequency")

plt.show()
```

```python
# Scatter plot

plt.scatter(air["Wind"], air["Temp"])

plt.title("Wind vs Temperature")

plt.xlabel("Wind")

plt.ylabel("Temperature")

plt.show()
```

```python
# Heatmap

sns.heatmap(air.corr(numeric_only=True), annot=True)

plt.title("Air Quality Heatmap")

plt.show()
```
