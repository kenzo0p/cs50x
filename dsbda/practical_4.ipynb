```python
import pandas as pd
```

```python
#data load
data = pd.read_csv("dataset_Facebook.csv", sep=';')

print(data.head())
```

```python
#create data subset
subset = data[["Page total likes", "Type", "like", "comment", "share"]]

print(subset.head())
```

```python
#merge data
data1 = data[["Page total likes", "Type"]]

data2 = data[["like", "comment", "share"]]

merged_data = pd.concat([data1, data2], axis=1)

print(merged_data.head())
```

```python
#sort data
sorted_data = data.sort_values(by="like", ascending=False)

print(sorted_data[["Type", "like"]].head())
```

```python
#transpose data
transpose_data = data.head().T

print(transpose_data)
```

```python
#reshape data
reshaped_data = data[["like", "comment", "share"]].values.reshape(-1, 3)

print(reshaped_data)
```
